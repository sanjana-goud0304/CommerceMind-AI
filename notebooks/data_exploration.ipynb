```python
master_df = (
    orders_payments.merge(
        order_items,
        on="order_id",
        how="inner"
    )
)
```

```python
#Revenue by Product Category
category_revenue = (master_df.groupby('product_category_name')['payment_value']
                    .sum().
                    sort_values(ascending=False)
                    .head(10)
)
category_revenue
```

```python
category_revenue.plot(kind="bar")

plt.title("Top Categories by Revenue")

plt.xlabel("Category")

plt.ylabel("Revenue")

plt.show()
```

```python
master_df = master_df.merge(
    customers,
    on="customer_id",
    how="inner"
)

master_df.head()
```

```python
master_df.shape
```

```python
master_df.columns
```

```python
#Customer Total Spending
customer_spending = (
    master_df.groupby(
        "customer_unique_id"
    )["payment_value"]
    .sum()
    .sort_values(ascending=False)
)

customer_spending.head()
```

```python
#Customer Order Frequency
customer_orders = (
    master_df.groupby(
        "customer_unique_id"
    )["order_id"]
    .nunique()
    .sort_values(ascending=False)
)
customer_orders.head()
```

```python
#Create Customer Segments
customer_summary = (
    master_df.groupby("customer_unique_id")
    .agg({
        "payment_value": "sum",
        "order_id": "nunique"
    })
    .reset_index()
)
customer_summary.columns = ["customer_id", "total_spent", "total_orders"]
customer_summary.head()
```

```python
customer_summary.shape
```

```python
def segment_customer(spent):

    if spent > 5000:
        return "VIP"

    elif spent > 2000:
        return "Premium"

    else:
        return "Regular"
```

```python
customer_summary["segment"] = (
    customer_summary["total_spent"]
    .apply(segment_customer)
)

customer_summary.head()
```

```python
customer_summary["segment"]\
    .value_counts()\
    .plot(kind="bar")

plt.title("Customer Segments")

plt.xlabel("Segment")

plt.ylabel("Customers")

plt.show()
```
